```python
import os
import torch
import joblib
import pandas as pd
import numpy as np
from sklearn.metrics import accuracy_score, classification_report
from transformers import DistilBertForSequenceClassification, DistilBertTokenizerFast
```

```python
tokenizer = DistilBertTokenizerFast.from_pretrained("files/tokenizer")
global_label_encoder = joblib.load("files/label_encoder.pkl")
```

```python
num_labels = len(global_label_encoder.classes_)
print(f"Disease classes: {num_labels}")
```

```python
# Load from a locally saved DistilBERT directory if present, otherwise
# download distilbert-base-uncased from HuggingFace Hub (architecture only;
# weights are then overwritten by the federated global_model1.pt).
local_dir = "files/distilbert_local"
base = local_dir if os.path.isdir(local_dir) else "distilbert-base-uncased"

global_model = DistilBertForSequenceClassification.from_pretrained(base, num_labels=num_labels)
global_model.load_state_dict(
    torch.load("files/global_model1.pt", map_location=torch.device("cpu"))
)
global_model.eval()
print("Global model loaded.")
```

```python
df=pd.read_csv('gretalai_test.csv')
df.head()
```

```python
device = torch.device("cpu")
global_model.to(device)
```

```python
device = torch.device("cpu")
global_model.to(device)

# Single-sample sanity check
test_text = str(df['description'].iloc[0])
with torch.no_grad():
    inputs = tokenizer(
        test_text, return_tensors="pt",
        padding=True, truncation=True, max_length=256,
    ).to(device)
    logits = global_model(**inputs).logits

pred_label  = torch.argmax(logits, dim=1).item()
pred_disease = global_label_encoder.inverse_transform([pred_label])[0]
actual       = df['disease'].iloc[0]

print(f"Input   : {test_text[:110]}...")
print(f"Predicted : {pred_disease}")
print(f"Actual    : {actual}")
print(f"Correct   : {pred_disease == actual}")
```

```python
# Full dataset evaluation — accuracy + weighted F1 + per-class breakdown
known_mask = df['disease'].isin(global_label_encoder.classes_)
eval_df    = df[known_mask].reset_index(drop=True)
skipped    = (~known_mask).sum()
if skipped:
    print(f"Skipped {skipped} rows with unseen disease labels.")

true_labels = global_label_encoder.transform(eval_df['disease'].values)
preds       = []

with torch.no_grad():
    for text in eval_df['description']:
        inputs = tokenizer(
            str(text), return_tensors="pt",
            padding=True, truncation=True, max_length=256,
        ).to(device)
        logits = global_model(**inputs).logits
        preds.append(torch.argmax(logits, dim=1).item())

acc = accuracy_score(true_labels, preds)
print(f"Test Accuracy : {acc:.4f}\n")
print(classification_report(
    true_labels, preds,
    target_names=global_label_encoder.classes_,
    zero_division=0,
))
```

```python
import shap
```

```python
def explain_prediction(model, tokenizer, text, label_encoder):
    if isinstance(text, pd.Series):
        text = text.astype(str).tolist()
    elif isinstance(text, str):
        text = [text]

    id2label = {i: label for i, label in enumerate(label_encoder.classes_)}

    def model_wrapper(input_texts):
        if isinstance(input_texts, np.ndarray):
            input_texts = input_texts.tolist()
        elif isinstance(input_texts, str):
            input_texts = [input_texts]
        elif not isinstance(input_texts, list) or not all(isinstance(t, str) for t in input_texts):
            raise ValueError("Input must be a string or a list of strings.")

        tokenized = tokenizer(input_texts, return_tensors="pt", padding=True, truncation=True, max_length=256).to(device)
        with torch.no_grad():
            logits = model(**tokenized).logits
        return logits.cpu().numpy()

    masker = shap.maskers.Text(tokenizer)
    explainer = shap.Explainer(model_wrapper, masker)
    shap_values = explainer(text)

    shap_values.output_names = [id2label[i] for i in range(len(id2label))]

    shap.text_plot(shap_values)
```

```python
explain_prediction(global_model, tokenizer, test_text, global_label_encoder)
```
